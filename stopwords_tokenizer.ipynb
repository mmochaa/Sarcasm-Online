```python
# Use of Stop-words and Tokenization while detecting sarcasm
# Hard-coded subreddit, for simplicity and to demonstrate concepts learned in class
# Program author: Mallika S.

import re 
import string

# A list of common Stopwords:
stop_words = { 
    "the","and","is","in","at","a","an","of","for","on","to","with",
    "it","this","that","but","so","are","was","be","or","as","if",
    "i","you","he","she","they","we","my","your","our","their","me"
}

punct = set(string.punctuation)

# Simple Tokenizer, using regex:
def tokenize(text):
    # Splits words, removes punctuation and converts to lowercase
    tokens = re.findall(r"\b\w+\b", text.lower())
    return tokens 

# Removal of stopwords
def remove_stopwords(tokens):
    return [t for t in tokens if t not in stop_words]

# Simple list of common saracasm keywords
sarcasm_keywords = { 
    "totally","great","amazing","yeah","sure","genius","nailed","loved","intentional","masterchef"
}

def detect_sarcasm(tokens): 
    return any(word in sarcasm_keywords for word in tokens)

# the mock subreddit "r/CookingFails" (created here itself, just for this project)   
# It has 15 comments, for simplicity purposes

comments = [
    "Wow, great job burning water. You're a culinary genius.",
    "I tried making pasta and it literally turned into soup.",
    "Totally didn't set off the smoke alarm again.",
    "Amazing! Another blackened pizza. Just what I wanted.",
    "My soufflé collapsed but at least the dog liked it.",
    "Sure, because adding more salt ALWAYS fixes everything.",
    "I baked cookies and they came out like tiny rocks.",
    "Oh yeah, the fire department loved my cooking experiment!",
    "I accidentally used sugar instead of salt. Masterchef material.",
    "The cake was supposed to be fluffy, not a brick.",
    "Totally intentional that my omelet became scrambled eggs.",
    "Burned the toast again. I'm on a roll!",
    "Yeah, because smoke flavor is totally the trend now.",
    "Made soup but forgot the broth. Nailed it.",
    "Nothing says 'success' like a melted plastic spatula."
]

# Analyzing the above subreddit comments: 
print("==== Sarcasm Detection in r/CookingFails ====\n")

for i, text in enumerate(comments):
    tokens = tokenize(text)
    filtered = remove_stopwords(tokens)
    sarcastic = detect_sarcasm(filtered)

    print(f"Comment {i+1}: {text}")
    print("Tokens:", filtered)
    print("Sarcastic? →", "YES" if sarcastic else "NO")
    print()
```
