```python
# Big thanks to Geeks for Geeks and W3Schools for information regarding Pandas

# CSV to dataframe
import pandas as pd
df = pd.read_csv('train-balanced-sarcasm.csv')
print(df.head())
```

```python
selected_columns = ['subreddit', 'score']
df_simplified = df[selected_columns].copy()
# alternatively df_simplified[['subreddit', 'score']] = df[['subreddit', 'score']]

data = {
    'subreddit': df['subreddit'].unique(),
    'average_score': df_simplified.groupby('subreddit')['score'].mean().values,
    'variance': df_simplified.groupby('subreddit')['score'].var().values,
    'standard_dev': df_simplified.groupby('subreddit')['score'].std().values,
    'count': df_simplified.groupby('subreddit')['score'].count().values
}
score_analysis_df = pd.DataFrame(data)

print(score_analysis_df.head(20))

# NOTE: Currently has all comments. ONLY HAVE ITEMS FROM label: 1
# This is also wrong: Politics has more than one entry among others
```

```python
# Gathering only sarcastic comments
selected_columns = ['subreddit', 'score']
dfsarcastic = df[df['label'] == 1].copy()
# dfsarcastic.to_csv("df2test.csv", index=False)
# print(dfsarcastic.head(10))

dfsarc_simp = dfsarcastic[selected_columns].copy()
# print(dfsarc_simp .head(10))

# Adjust 'y' in len(x) > y to filter what amount of comments had to be scraped from a given subreddit to be considered
groupedsarc_df = dfsarc_simp.groupby('subreddit').filter(lambda x: len(x) > 1)
# print(groupedsarc_df)
# print(groupedsarc_df.groupby('subreddit').describe())
descsarc_df = groupedsarc_df.groupby('subreddit').describe()
print(descsarc_df)
```

```python
# Getting mean of all scores
print("Mean of sarcastic comments: ", dfsarcastic['score'].mean())

# Gathering only genuine comments
selected_columns = ['subreddit', 'score']
dfgenuine = df[df['label'] == 0].copy()
print("Mean of genuine comments: ", dfgenuine['score'].mean())
```

```python
# Mean vs Std of sarcastic comments
import matplotlib
matplotlib.use('qtagg')
import matplotlib.pyplot as plt

x = descsarc_df[('score', 'mean')]
y = descsarc_df[('score', 'std')]

# Create scatter plot
plt.figure(figsize=(8,6))
plt.scatter(x, y)
plt.xlabel("Mean")
plt.ylabel("Standard Deviation")
plt.title("Mean vs Std of Score")
plt.grid(True)
plt.show()
```

```python
# Mean vs std of genuine comments
import matplotlib
matplotlib.use('qtagg')
import matplotlib.pyplot as plt

# Creating genuine simplified dataset
dfgen_simp = dfgenuine[selected_columns].copy()
groupedgen_df = dfgen_simp.groupby('subreddit').filter(lambda x: len(x) > 1)
descgen_df = groupedgen_df.groupby('subreddit').describe()

x_gen = descgen_df[('score', 'mean')]
y_gen = descgen_df[('score', 'std')]
# Create scatter plot
plt.figure(figsize=(8,6))
plt.scatter(x_gen, y_gen)
plt.xlabel("Mean")
plt.ylabel("Standard Deviation")
plt.title("Mean vs Std of Genuine Comment Scores")
plt.grid(True)
plt.show()
```

```python
# Best received subreddits
top_mean_subreddits = descsarc_df[("score", "mean")].nlargest(10)
print(top_mean_subreddits)

# Bar graph
top_mean_subreddits.plot.bar()
plt.tight_layout()
plt.xlabel("Subreddit")
plt.ylabel("Mean")
plt.title("Highest Scores by Subreddit")
plt.show()
```

```python
# Worst received subreddits
bottom_mean_subreddits = descsarc_df[("score", "mean")].nsmallest(10)
print(bottom_mean_subreddits)

# Bar graph
bottom_mean_subreddits.plot.bar()
plt.tight_layout()
plt.xlabel("Subreddit")
plt.ylabel("Mean")
plt.title("Lowest Scores by Subreddit")
plt.show()
```
