```python
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
import matplotlib.pyplot as plt
from collections import Counter


df = pd.read_csv(r"c:\Users\micha\Documents\Datamining\train-balanced-sarcasm.csv\train-balanced-sarcasm.csv")


def get_pos_tags(text):
    tokens = word_tokenize(str(text))
    return pos_tag(tokens)

pos_counts = Counter()

for comment in df["comment"].dropna():
    tagged = get_pos_tags(comment)
    for _, tag in tagged:
        pos_counts[tag] += 1


pos_df = pd.DataFrame.from_dict(pos_counts, orient="index", columns=["count"])
pos_df = pos_df.sort_values("count", ascending=False)

print(pos_df)


plt.figure(figsize=(12, 6))
plt.bar(pos_df.index, pos_df["count"], color="steelblue")
plt.title("POS Tag Distribution")
plt.xlabel("POS Tag")
plt.ylabel("Frequency")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
